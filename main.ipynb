```python
import working_model as model
import numpy as np
import matplotlib.pyplot as plt
import pandas

from sklearn.model_selection import train_test_split
```

```python
inputVectors = np.array(
    [
        [3, 1.5],
        [2, 1],
        [4, 1.5],
        [3, 4],
        [3.5, 0.5],
        [2, 0.5],
        [5.5, 1],
        [1, 1],
    ]
 )

targets = np.array([0, 1, 0, 1, 0, 1, 1, 0])
```

```python
df = pandas.read_csv('Data/exo01.txt',sep=',')
X = df[['x1', 'x2']]
y = df['class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=1)
```

```python
x1_train = X_train["x1"].to_list()
x2_train = X_train["x2"].to_list()
x1_test = X_test["x1"].to_list()
x2_test = X_test["x2"].to_list()

inputVectors = []
testInputVectors = []

for i in range(0, len(x1_train)):
    inputVectors.append([x1_train[i], x2_train[i]])
    
for i in range(0, len(x1_test)):
    testInputVectors.append([x1_test[i], x2_test[i]])
    
targets = y_train.to_list()
testTargets = y_test.to_list()
```

```python
learning_rate = 0.1
iterations = 1000
neural_network = model.MyFirstNeuralNetwork(learning_rate, 2, len(inputVectors[0]))

trainingError, testError = neural_network.train(inputVectors, targets, iterations, testInputVectors, testTargets)
print(neural_network.getWeights())
print(neural_network.getBiases())

#plt.savefig("cumulative_error.png")
```

```python
size = len(testInputVectors)
print(testInputVectors[np.random.randint(size)])
print(testTargets[np.random.randint(size)])
neural_network.predict(testInputVectors[np.random.randint(size)])
```

```python
plt.plot(trainingError)
plt.plot(testError)
plt.xlabel("Iterations")
plt.ylabel("Error for all training instances")
```
